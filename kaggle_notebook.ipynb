# Pokémon Battle Prediction — FDS Kaggle Competition

This notebook clones our GitHub repository and reproduces our 3 final submissions.
All the code lives in the repo (`src/`); this notebook only calls it.

```python
# 1) Clone the GitHub repo (CHANGE the URL to yours)
import os, sys, subprocess

REPO_URL = "https://github.com/YOUR-USERNAME/fds-pokemon.git"
REPO_DIR = "/kaggle/working/fds-pokemon"

# While the repo is private, add a Kaggle secret called GITHUB_TOKEN (Add-ons > Secrets).
try:
    from kaggle_secrets import UserSecretsClient
    token = UserSecretsClient().get_secret("GITHUB_TOKEN")
    REPO_URL = REPO_URL.replace("https://", f"https://{token}@")
except Exception:
    pass  # public repo: no token needed

if not os.path.exists(REPO_DIR):
    subprocess.run(["git", "clone", "--depth", "1", REPO_URL, REPO_DIR], check=True)
sys.path.insert(0, REPO_DIR)
print("repo ready")
```

```python
# 2) Load data and build features (shared by the 3 submissions)
from src.data import load_data, get_target
from src.features import build_features
from src.models import get_model
from src.validation import evaluate
from src.submission import make_submission

train, test, sample = load_data()
X_train, y = build_features(train), get_target(train)
X_test = build_features(test)[X_train.columns]
print(X_train.shape, X_test.shape)
```

## Submission 1 — Logistic regression (baseline)

*Explain the approach here: features used, model, CV accuracy.*

```python
model = get_model("logreg")
evaluate(model, X_train, y)
model.fit(X_train, y)
make_submission(model, X_test, "submission_1.csv", sample)
```

## Submission 2 — Random forest

*Explain the approach here.*

```python
model = get_model("rf")
evaluate(model, X_train, y)
model.fit(X_train, y)
make_submission(model, X_test, "submission_2.csv", sample)
```

## Submission 3 — Gradient boosting (XGBoost)

*Explain the approach here.*

```python
model = get_model("xgb")
evaluate(model, X_train, y)
model.fit(X_train, y)
make_submission(model, X_test, "submission_3.csv", sample)
```
